# EV Market & Sales Analytics — Business Insights

## 1. Project Overview

This project analyzes 3,022 electric vehicle records to understand 2024 recorded sales performance, manufacturer and model performance, vehicle specifications, charging technologies, and manufacturing patterns.

## 2. Key Business Insights

### Manufacturer Performance
- Ferrari recorded the highest total 2024 units sold among manufacturers in the dataset.
- VinFast, Nissan, Toyota, and Great Wall Motors also recorded high sales volumes.

### Model Performance
- Rimac Nevera recorded the highest units sold among the EV models analyzed.
- Jaguar I-PACE and Faraday Future FF 91 were also among the highest-selling models.

### Price and Sales Relationship
- Price vs Sales correlation was **-0.013**.
- This indicates an almost negligible linear relationship between price and recorded sales in this dataset.

### Range and Sales Relationship
- Range vs Sales correlation was **0.014**.
- This indicates an almost negligible linear relationship between driving range and recorded sales.

### Battery and Charging Technology
- The dataset contains multiple battery technologies and charging types.
- This provides an opportunity to compare EV technologies across the vehicles analyzed.

### Manufacturing Market
- EVs in the dataset are manufactured across multiple countries.
- Country-level analysis helps understand the geographical distribution represented in the dataset.

### Safety and Autonomous Features
- Safety ratings and autonomous driving levels vary across the dataset.
- Comparing these features with recorded sales helps identify potential relationships.

## 3. Business Recommendations

- Monitor high-performing manufacturers and models for further investigation.
- Avoid relying on vehicle price alone when analyzing sales performance.
- Evaluate battery, range, charging, safety, and autonomous features together.
- Track the distribution of battery and charging technologies.
- Use country-level analysis to identify manufacturing patterns.

## 4. Data Limitation

The analysis is based on the provided dataset and its recorded `Units_Sold_2024` values. These figures should not automatically be interpreted as verified real-world market sales.

Correlation results describe relationships within this dataset and do not establish causation.

## 5. Conclusion

The analysis provides a structured view of EV product characteristics and recorded 2024 sales performance. Manufacturer, model, technology, country, safety, and vehicle-feature analysis can support further business investigation and decision-making.
